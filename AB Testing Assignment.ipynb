```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```
data = pd.read_csv('/content/ab_data.csv')
```

```
# Check for missing values
print(data.isnull().sum())

# Drop duplicates if necessary
data = data.drop_duplicates()
```

```
# Filter control group (assuming it's called 'control' in 'group' column)
control_group = data[data['group'] == 'control']

# Filter treatment group (assuming it's called 'treatment' in 'group' column)
treatment_group = data[data['group'] == 'treatment']
```

```
from statsmodels.stats.proportion import proportions_ztest
from statsmodels.stats.weightstats import ztest

```

```
# Define the success metrics (conversion rates)
control_conversion_rate = control_group['converted'].mean()
treatment_conversion_rate = treatment_group['converted'].mean()

# Calculate the sample size and standard error
n_control = len(control_group)
n_treatment = len(treatment_group)

# Perform the Z-test
z_score, p_value = proportions_ztest([treatment_group['converted'].sum(), control_group['converted'].sum()],
                                           [n_treatment, n_control])

print(f'Z-Score: {z_score}')
print(f'P-Value: {p_value}')
```
