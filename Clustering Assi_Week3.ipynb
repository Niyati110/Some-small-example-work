```
from sklearn import cluster
```

```
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```
all_data = pd.read_excel('/content/testforclusternew(1).xlsx')
```

```
num_data = all_data.iloc[:,2:8]
```

```
k_clusters=3
```

```
labels = cluster.KMeans(n_clusters = k_clusters).fit(num_data).labels_
```

```
all_data['Cluster Assignment'] = labels
```

```
print(all_data)
```

```
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(num_data.iloc[:, 0], num_data.iloc[:, 1], num_data.iloc[:, 2],
                     c=labels, cmap='viridis')
```

```
# Adding labels and titles
ax.set_xlabel('Unnamed: 2')
ax.set_ylabel('Unnamed: 3')
ax.set_zlabel('Unnamed: 4')
ax.set_title('3D Cluster Visualization')
```

```
legend1 = ax.legend(*scatter.legend_elements(), title="Clusters")
ax.add_artist(legend1)
```

```
plt.show()
```

```
# @title Observation
# Observation:-  In the clustering result,Cluster 0 comprises data that has moderate levels for `Unnamed: 2` and mid-levels for variable `Unnamed: 4`, while Cluster 1 has relatively high levels `Unnamed: 3` values with lower `Unnamed: 4` values. Cluster 2 has higher levels of `Unnamed: 2` and mixed levels for the other two variables in the cluster.
```
